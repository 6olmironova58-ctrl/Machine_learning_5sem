````python
import pandas as pd
````

````python
df = pd.read_csv('russian-housing-market.csv')
````

1. Посчитайте количество записей в датасете с отсутствующей информацией о жилой площади жилого помещения

````python
print(df['life_sq'].isna().sum())
````

2. Предложите способ заменить отсутствующие значения из пункта 1 и реализуйте его. Способ нужно аргументировать(сказать почему это достаточно хорошее приближение)

````python
# Считаем медианное соотношение жилой площади к общей
# отражает как статистически относится жилая к общей, позволяет заполнить пропуски логичне, 
# чем придумывая процент основываясь на личном жизненном опыте
valid_ratio = (df.loc[df['full_sq'] > 0, 'life_sq'] / df.loc[df['full_sq'] > 0, 'full_sq']).median()

result_df = df.copy()
result_df['life_sq'] = result_df['life_sq'].fillna(result_df['full_sq'] * valid_ratio)

print(f"Медианное соотношение life_sq к full_sq: {valid_ratio:.2f}")
print(f"Осталось пропусков: {result_df['life_sq'].isna().sum()}")
````

4. Посчитайте среднюю стоимость продажи помещений с общей площадью от 50 до 70 квадратных метров

5. Какая максимальная этажность здания представлена в датасете

````python
print(df['max_floor'].fillna(df['floor']).max())
````

6. Выявите район с самым дорогим жильем

````python
# самое дорогое жилье там, где самая высокая цена за квадратный метр
df['price_per_m'] = np.where(
    df['full_sq'] > 0, 
    df['price_doc'] / df['full_sq'], 
    np.nan
)
avg_price_per_m = df.groupby('sub_area')['price_per_m'].mean().sort_values(ascending=False)
print("Самая высокая цена за кв метр:", avg_price_per_m.index[0])

# или если площадь не важна, а важна только цена
avg_price = df.groupby('sub_area')['price_doc'].mean().sort_values(ascending=False)
print("Самые дорогие квартиры:", avg_price.index[0])
````

7. Посчитайте среднюю стоимость за квадратный метр жилья в Митино(Mitino)

8. Посчитайте среднюю площадь комнаты в Бибирево

````python
bibirevo = df[(df["sub_area"] == "Bibirevo") & (df["num_room"].notna())].copy()
bibirevo['room_area'] = bibirevo['life_sq'] / bibirevo['num_room']
print(bibirevo['room_area'].mean())
````

9. Проанализируйте есть ли разница в цене между квартирами на 1 и 9 этажах

````python
# df['price_per_m'] сделали в 6 номере
price_m_floor_1 = df[df['floor'] == 1]['price_per_m'].mean()
price_m_floor_9 = df[df['floor'] == 9]['price_per_m'].mean()

print(f"Средняя цена за м² на 1 этаже: {price_m_floor_1:,.2f} руб.")
print(f"Средняя цена за м² на 9 этаже: {price_m_floor_9:,.2f} руб.")

diff = price_m_floor_9 - price_m_floor_1
if diff > 0:
    print(f"Квартиры на 9 этаже в среднем дороже")
elif diff < 0:
    print(f"Квартиры на 1 этаже в среднем дороже")
else:
    print("Цены за м² на 1 и 9 этажах статистически одинаковы")
````

### 3. Визуализация

Используя библиотеки matplotlib/sns/plotly/pandas и набор данных представленный ранее визуализируйте

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
````

1. Как менялась средняя цена квартиры в Нагатинском Затоне(Nagatinski Zaton) с течением времени

````python
nagat = df[df['sub_area'] == 'Nagatinskij Zaton'].copy()

nagat['date'] = pd.to_datetime(nagat['timestamp'])
nagat['year_month'] = nagat['date'].dt.to_period('M')

price_by_month = nagat.groupby('year_month')['price_doc'].mean() / 1_000_000

plt.plot(price_by_month.index.astype(str), price_by_month.values, marker='o', linewidth=2)
plt.title('Средняя цена квартиры в Нагатинском Затоне по месяцам')
plt.xlabel('Месяц')
plt.ylabel('Средняя цена (млн руб)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````

2. Постройте гистограмму распределения цен на квартиры с общей площадью более 50 кв.метров

````python
df_50 = df[(df['full_sq'] > 50) & (df['price_doc'] < 50_000_000)]['price_doc'] / 1_000_000

plt.hist(df_50, bins=50, color='skyblue', edgecolor='black')
plt.title('Распределение цен на квартиры (площадь > 50 кв.м)')
plt.xlabel('Цена (млн руб)')
plt.ylabel('Количество квартир')
plt.tight_layout()
plt.show()
````

3. Визуализируйте количество проданных квартир в разных районах

````python
top_districts = df['sub_area'].value_counts().head(15).sort_values(ascending=True)

plt.figure(figsize=(10, 7))
plt.barh(top_districts.index, top_districts.values, color='green', edgecolor='black')
plt.title('Топ-15 районов по количеству проданных квартир')
plt.xlabel('Количество сделок')
plt.ylabel('Район')
plt.tight_layout()
plt.show()
````

4. Постройте box plot отображающий распределение полных площадей квартир в Басманном районе

````python
basmannoe = df[df['sub_area'] == 'Basmannoe']['full_sq']

plt.figure(figsize=(6, 8))
plt.boxplot(basmannoe, patch_artist=True, 
            boxprops=dict(facecolor='lightblue', color='blue'),
            medianprops=dict(color='red', linewidth=2))

plt.title('Распределение полных площадей квартир в Басманном районе')
plt.ylabel('Площадь (кв.м)')
plt.xticks([1], ['Basmannoe'])

plt.show()
````

5. Визуализируйте зависимость средней стоимости квартир в зависимости от этажа

````python
floor_price = df.groupby('floor')['price_doc'].mean().reset_index()

floor_price = floor_price[(floor_price['floor'] >= 1) & (floor_price['floor'] <= 25)]
floor_price['price_doc'] = floor_price['price_doc'] / 1_000_000

plt.plot(floor_price['floor'], floor_price['price_doc'], marker='s', color='orange', linewidth=2)
plt.title('Средняя цена квартиры в зависимости от этажа (до 25 этажа)')
plt.xlabel('Этаж')
plt.ylabel('Средняя цена (млн руб)')
plt.xticks(range(1, 26))
plt.tight_layout()
plt.show()
````

6. Покажите как меняется средняя площадь квартир в зависимости от района

````python
district_area = (df.groupby('sub_area')['full_sq'].mean()
                 .sort_values(ascending=False)
                 .head(20)
                 .sort_values())

plt.barh(district_area.index, district_area.values, color='pink', edgecolor='black')
plt.title('Топ-20 районов по средней площади квартиры')
plt.xlabel('Средняя площадь (кв.м)')
plt.ylabel('Район')
plt.tight_layout()
plt.show()
````

7. Постройте график изменения медианной цены на квартиру менее 50 кв. метров за период с 2011 до 2014 года*

\* Задание на доп. балл

````python
df['timestamp'] = pd.to_datetime(df['timestamp'])

clean_df = df[
    (df["full_sq"] < 50) &
    (df["timestamp"] >= "2011-01-01") &
    (df["timestamp"] <= "2014-12-31")
]

data = clean_df.groupby(clean_df["timestamp"].dt.to_period("M"))["price_doc"].median() / 1_000_000

plt.figure(figsize=(12, 6))
plt.plot(data.index.astype(str), data.values, marker='*', color='purple', linewidth=2, markersize=10)
plt.title('Медианная цена квартир < 50 кв.м (2011–2014)')
plt.xlabel('Год и месяц')
plt.ylabel('Медианная цена (млн руб)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````
